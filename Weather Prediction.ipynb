```python
#Import Eseential Libraries
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
```

```python
# Import Dataset
df=pd.read_csv("wp_data.csv")
df.head()
```

```python
#  Dropping date column as it's of no use 
df=df.drop("date",axis=1)
df.head()
```

```python
df.shape
```

```python
#Check for missing values
df.isnull().sum()
```

```python
#Separate dataset into dependent and independent variables
X=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
#train_test_splitting
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test=train_test_split(X, y, test_size=0.2,random_state=0)
```

```python
# Apply Logistic Regression Algorithm
from sklearn.linear_model import LogisticRegression
wp=LogisticRegression()
wp.fit(X_train, y_train)
```

```python
# Test the Model
y_pred=wp.predict(X_test)
y_pred
```

```python
y_test
```

```python
# Create confusion matrix
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test, y_pred)
print(cm) #237 matched 56 mismatched total Sample = 293
```

```python
from sklearn.metrics import accuracy_score
print(f"Accuracy Score ={accuracy_score(y_pred, y_test):.2%}")
```

```python
import pickle
pickle.dump(wp,open("wp.pkl","wb"))
print ("Model is saved")
```
